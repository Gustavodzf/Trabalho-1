```
movie-recommender.ipynb
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.neighbors import NearestNeighbors

# 🔹 Upload da base pelo usuário
# No Google Colab, use:
# from google.colab import files
# uploaded = files.upload()
# file_name = list(uploaded.keys())[0]

# No Jupyter local ou Codespaces, peça o nome do arquivo:
file_name = input("Digite o nome do arquivo CSV que deseja usar: ")

# Carregar os dados
df = pd.read_csv(file_name)

# 🔹 Vetorização
count = CountVectorizer(stop_words='english')
count_matrix = count.fit_transform(df['soup'].fillna(""))

# 🔹 Similaridade de Cosseno
cosine_sim = cosine_similarity(count_matrix, count_matrix)

def recomendar_cosine(titulo):
    try:
        idx = df[df['title'] == titulo].index[0]
        sim_scores = list(enumerate(cosine_sim[idx]))
        sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
        top_indices = [i[0] for i in sim_scores[1:6]]
        print(f"\n🎬 Recomendações (Cosine) para '{titulo}':")
        for i in top_indices:
            print(f"- {df['title'].iloc[i]}")
    except:
        print(f"\n❌ Filme '{titulo}' não encontrado.")

# 🔹 KNN
knn = NearestNeighbors(n_neighbors=6, metric='cosine')
knn.fit(count_matrix)

def recomendar_knn(titulo):
    try:
        idx = df[df['title'] == titulo].index[0]
        distances, indices = knn.kneighbors(count_matrix[idx])
        top_indices = indices[0][1:]
        print(f"\n🎬 Recomendações (KNN) para '{titulo}':")
        for i in top_indices:
            print(f"- {df['title'].iloc[i]}")
    except:
        print(f"\n❌ Filme '{titulo}' não encontrado.")

# 🔎 Testes
recomendar_cosine("Avengers: Age of Ultron")
recomendar_knn("Avengers: Age of Ultron")

recomendar_cosine("A Christmas Carol")
recomendar_knn("A Christmas Carol")
```
